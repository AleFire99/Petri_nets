# Coloured nets

In a coloured net, tokens carry data (here integers) and arcs/transitions use expressions and guards. `petrilab.petri.colored.jobs_net` is built directly with SNAKES: `square` consumes a job `n` from `Jobs` under guard `n > 1` and produces `n*n` in `Results`.

```mermaid
flowchart LR
    Jobs((Jobs 1,2,3)) -- n --> square["square [n > 1]"]
    square -- "n*n" --> Results((Results))
```

````python
from petrilab.petri import analysis
from petrilab.petri.colored import jobs_net
from petrilab.petri.snakes_bridge import explore

space = explore(jobs_net())
print("states:", len(space))
(dead,) = analysis.dead_markings(space)
final = space.markings[dead]
print("Jobs:", sorted(final["Jobs"]), "Results:", sorted(final["Results"]))
````

Jobs 2 and 3 are squared (giving 4 and 9); job 1 fails the guard `n > 1` and stays in `Jobs` forever. There are 4 reachable markings (the subsets of processed jobs) and exactly one dead marking.

````python
space = explore(jobs_net((1,)))
print("states:", len(space), "fired:", analysis.fired_transitions(space))
````

With only job `1`, the guard is never satisfied: a single state and no transition ever fires.

````python
space = explore(jobs_net((2, 3, 4)))
(dead,) = analysis.dead_markings(space)
print(sorted(space.markings[dead]["Results"]), space.tokens(dead, "Jobs"))
````

When every job satisfies the guard all are processed: `Results = [4, 9, 16]` and `Jobs` is empty in the final marking.
